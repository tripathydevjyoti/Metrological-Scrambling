Import Libraries


```python
import numpy as np
import matplotlib.pyplot as plt
from qutip import basis, tensor, sigmax, sigmaz, sigmay, qeye, ptrace, Qobj
from scipy.linalg import lu_factor, lu_solve
from scipy.sparse import csc_matrix
from scipy.sparse.linalg import eigsh
from mpl_toolkits.mplot3d import Axes3D 
import plotly.graph_objects as go
from scipy.integrate import cumulative_trapezoid
```

## Crank-Nicolson Unitary Propagator

```python
def crank_nicolson_propagator(H: Qobj, t: float, steps: int = 100) -> Qobj:
    dt = t / steps
    dim = H.shape[0]
    I = np.eye(dim, dtype=complex)
    A = I - 1j * (dt/2) * H.full()
    B = I + 1j * (dt/2) * H.full()

    lu, piv = lu_factor(A)
    U_step = lu_solve((lu, piv), B)
    U_total = np.linalg.matrix_power(U_step, steps)
    
    return Qobj(U_total, dims=H.dims)
```

## Spin Chain Hamiltonian

```python
def spin_chain_ham(h: float, g: float, J: float, J2: float, N: int):
    sx = [tensor([sigmax() if i == j else qeye(2) for i in range(N)]) for j in range(N)]
    sz = [tensor([sigmaz() if i == j else qeye(2) for i in range(N)]) for j in range(N)]
    
    H = sum(-J * sz[i] * sz[i+1] for i in range(N-1)) \
        + sum(-J2 *sz[i] * sz[i+2] for i in range(N-2)) \
        - sum(h * sx[i] for i in range(N))\
        - sum(g * sz[i] for i in range(N))
    return H
```

## Subsystem Quantum Fisher Information

```python
def compute_time_qfi_spectral(states, tlist, subsystem_index=0):
    qfi_list = []

    for i in range(len(tlist)):
        rho_A = ptrace(states[i], [subsystem_index])

        # Finite difference derivative
        if i == 0:
            rho_A_next = ptrace(states[i+1], [subsystem_index])
            dt = tlist[i+1] - tlist[i]
            drho_dt = (rho_A_next - rho_A) / dt
        elif i == len(tlist)-1:
            rho_A_prev = ptrace(states[i-1], [subsystem_index])
            dt = tlist[i] - tlist[i-1]
            drho_dt = (rho_A - rho_A_prev) / dt
        else:
            rho_A_next = ptrace(states[i+1], [subsystem_index])
            rho_A_prev = ptrace(states[i-1], [subsystem_index])
            dt = tlist[i+1] - tlist[i-1]
            drho_dt = (rho_A_next - rho_A_prev) / dt

        # Eigendecomposition
        evals, evecs = rho_A.eigenstates()
        qfi_t = 0.0

        for m in range(len(evals)):
            for n in range(len(evals)):
                pm, pn = evals[m], evals[n]
                if pm + pn > 1e-12:
                    
                    # ---------------------------------------------
                    # FIX: convert Qobj 1×1 into python complex
                    # ---------------------------------------------
                    d_mn = complex(evecs[m].dag() * drho_dt * evecs[n])

                    qfi_t += (2.0 / (pm + pn)) * abs(d_mn)**2

        qfi_list.append(np.real(qfi_t))

    return np.array(qfi_list)
```

## Subsystem QFI for different values of h

```python
def run_simulation_for_h(h_value):
    print(f"Running simulation for h = {h_value}")

    H_full = spin_chain_ham(h_value, g, J, J2, N)
    H_init_full = spin_chain_ham(h_value, g, J, J2, N)
    H_sparse_full = csc_matrix(H_init_full.full())
    E_full, psi_full = eigsh(H_sparse_full, k=1, which='SA')   # smallest algebraic
    psi0_full = Qobj(psi_full[:, 0], dims=[H_full.dims[0], [1]*N])
    #psi0_full = tensor([basis(2, 0) for _ in range(N)])
    #psi0_full = tensor([basis(2, i % 2) for i in range(N)])

    # Pure initial ground state density matrix
    rho0 = psi0_full * psi0_full.dag()



    # Local quench operator on first spin
    O = tensor(sigmax(), *[qeye(2) for _ in range(N-1)])

    # Apply quench
    initial_state = O * rho0 * O.dag()

    states = []
    purity_t = []
    for t in tlist:
        
        U_full = crank_nicolson_propagator(H_full, t, steps)
        evolved_state = U_full * initial_state * U_full.dag()
        states.append(evolved_state)
        rho_A = ptrace(evolved_state, [0])
        rho_0 = ptrace(initial_state, [0])
        purity_t.append(np.real((rho_A * rho_A).tr()) )
        
    purity_h = np.array(purity_t)

    qfi_vals = compute_time_qfi_spectral(states, tlist, subsystem_index=0)
    return qfi_vals, purity_h
```


# Parameters (global)



```python
N = 6
J = 1.0
J2= 0.0
g = 0.4
#beta = 1.0
alpha = 2

h_list = np.linspace(-3.0, 3.0,20)

tlist = np.linspace(0, 1.0, 50)

steps = 100
```

# Subsystem QFI(h,t) 3D Plot

```python


qfi_results = {}
purity_results = {}
for h in h_list:
    qfi_results[h], purity_results[h] = run_simulation_for_h(h)
    
# Build QFI matrix: shape (len(h_list), len(tlist))
QFI_matrix = np.array([qfi_results[h] for h in h_list])

    

fig = go.Figure(data=[go.Surface(
    x=h_list,
    y=tlist,
    z=QFI_matrix.T,
    colorscale='Viridis'
)])

fig.update_layout(
    title="Interactive 3D QFI(h, t)",
    scene=dict(
        xaxis_title='h',
        yaxis_title='Time t (estimation parameter)',
        zaxis_title='Subsystem QFI'
    ),
    autosize=True,
    height=700
)

fig.show()

```

# Compute $\frac{\Big(\int_{0}^{t}\sqrt{I_{F}}ds\Big)^2}{8t}$

```python

F_results = {}


for h in h_list:
    qfi_t = np.array(qfi_results[h])
    qfi_t[qfi_t < 0] = 0.0

    # sqrt(QFI)
    sqrt_qfi = np.sqrt(qfi_t)

    # Z(t) = ∫₀ᵗ sqrt(QFI(s)) ds
    Z_t = cumulative_trapezoid(sqrt_qfi, tlist, initial=0)

    # Avoid division by zero at t = 0
    F_t = np.zeros_like(Z_t)
    
    
    F_t[1:] = (Z_t[1:]**2) / (8 * tlist[1:])
     

    F_results[h] = F_t





# Convert dict → matrix (h-indexed rows, t-indexed columns)
F_matrix = np.array([F_results[h] for h in h_list])

fig = go.Figure(
    data=[
        go.Surface(
            x=h_list,       # h-axis
            y=tlist,        # t-axis
            z=F_matrix.T,   # transpose → z[i,j] = F(h_i, t_j)
            colorscale="Viridis",
            showscale=True
        )
    ]
)

fig.update_layout(
    title=r"Interactive 3D Plot of  $F(h,t)=\dfrac{Z(t)^2}{8t}$",
    scene=dict(
        xaxis_title="h (Transverse Field)",
        yaxis_title="t (Time) (estimation parameter)",
        zaxis_title=r"$Z(t)^2/8t$",
        xaxis=dict(backgroundcolor="white"),
        yaxis=dict(backgroundcolor="white"),
        zaxis=dict(backgroundcolor="white")
    ),
    width=950,
    height=750,
)

fig.show()

```

```python
# =============================================================================
# Compute G(h,t) = - ln(purity) / t
# =============================================================================

G_results = {}

for h in h_list:
    purity_t = np.array(purity_results[h])

    # ensure purity is > 0
    purity_t[purity_t <= 0] = 1e-12

    G_t = np.zeros_like(purity_t)

    # avoid t=0 singularity
    G_t[1:] = -np.log(purity_t[1:]) / tlist[1:]

    G_results[h] = G_t
```

```python
# =============================================================================
# INTERACTIVE 3D PLOT OF -ln(purity)/t
# =============================================================================

import plotly.graph_objects as go
import numpy as np

G_matrix = np.array([G_results[h] for h in h_list])

fig = go.Figure(data=[
    go.Surface(
        x=h_list,
        y=tlist,
        z=G_matrix.T,
        colorscale="Turbo",
        showscale=True
    )
])

fig.update_layout(
    title=r"Interactive 3D Surface of  $-\ln(\mathrm{purity}) / t$",
    scene=dict(
        xaxis_title="h (transverse field)",
        yaxis_title="t (time)",
        zaxis_title=r"$-\ln(\gamma)/t$"
    ),
    autosize=True,
    width=900,
    height=700
)

fig.show()

```
